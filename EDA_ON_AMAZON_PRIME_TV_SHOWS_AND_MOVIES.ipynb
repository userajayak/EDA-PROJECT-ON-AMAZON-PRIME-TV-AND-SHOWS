## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Remove Duplicate Rows
titles.drop_duplicates(inplace= True)
credits.drop_duplicates(inplace=True)

# handle Missing Values

titles.replace("", np.nan, inplace=True)
credits.replace("", np.nan, inplace=True)

# Fill blank strings with NAN
titles.replace("", np.nan, inplace=True)
credits.replace("", np.nan, inplace= True)

# For Title dataset
titles['age_certification'].fillna("unknown, inplace=True")
titles['seasons'].fillna(0, inplace=True)
titles['genres'].fillna("[]", inplace=True)
titles['production_countries'].fillna("[], inplace=True")
titles['description'].fillna("No Description Available", inplace=True)

# Fill empty spaces with NAN and with median
num_cols = ['runtime', 'imdb_score',  'imdb_votes', 'tmdb_popularity', 'tmdb_score']
for col in num_cols:
  titles[col].fillna(titles[col].median(), inplace=True)

# Fill empty spaces with NAN in credits Dataset
credits['character'].fillna("unknown", inplace=True)

# Fix Incorrect Data
titles['release_year'] = titles['release_year'].astype(int)
titles['seasons'] = titles['seasons'].astype(int)



# Converting strings to python list ( JSON like columns)

import ast

def convert_to_list(x):
    try:
        return ast.literal_eval(x)

    except:

      return[]

titles['genres'] = titles['genres'].apply(convert_to_list)
titles['production_countries'] = titles['production_countries'].apply(convert_to_list)

# Stadardize Text columns

titles['title'] = titles['title'].str.strip()
credits['name'] = credits['name'].str.strip()

titles = titles[titles['runtime'] <= 500]
````

````python
# Chart - 1 visualization code

# TYPE VS GENRE

# BAR CHART

# 1. Explode genres into individual rows
df_genre = titles.explode('genres')

# 2. Identify the Top 10 genres by total frequency
top_10_genre_names = df_genre['genres'].value_counts().nlargest(10).index

# 3. Filter the dataframe to ONLY include those 10 genres
df_top_10 = df_genre[df_genre['genres'].isin(top_10_genre_names)]

# 4. Group by type and genre
type_genre_count = df_top_10.groupby(['type', 'genres']).size().reset_index(name='count')

# Plot
plt.figure(figsize=(12, 8))
sns.barplot(data=type_genre_count, x='genres', y='count', hue='type', order=top_10_genre_names)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.title('Chart 1: - Top 10 Genres by Type', fontsize=14)
plt.xlabel('Genre')
plt.ylabel('Count')
plt.legend(title='Type')
plt.tight_layout()
plt.show()
````

#### Chart - 3

````python
# Chart - 3 visualization code

# TOP 10 PRODUCTION COUNTRIES

# BAR CHART

df = pd.read_csv("/content/titles.csv")

def parse_countries(cell): # defining a function to covert
  try:                                           # string to PYTHON LIST
    if pd.isnull(cell):
      return[]
    parsed = ast.literal_eval(cell)
    if isinstance(parsed, list):  #  CHECK IF PARSED VALUE IS A PYTHON LIST.
      return[str(x).strip() for x in parsed if str(x).strip()]
    else:
      return[]
  except Exception:
    return[]

df['prod_countries_list'] = df['production_countries'].apply(parse_countries)

exploded = df.explode('prod_countries_list')
exploded = exploded[exploded['prod_countries_list'] !='']

#  This gives Top 10 producing countries
top_countries = exploded['prod_countries_list'].value_counts().head(10)
top10_df = top_countries.reset_index()
top10_df.columns = ['country', 'count']

#        PLOT CHART
plt.figure(figsize=(10,6))
plt.bar(top10_df['country'], top10_df['count'], alpha=0.9)
plt.xlabel('Country', fontsize=12)
plt.ylabel('Number of Titles', fontsize=12)
plt.title('chart - 3: Top 10 Production Countries', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

````python
# Chart - 4 visualization code

# RELEASE TREND: MOVIES VS TV SHOWS

# LINE CHART

df['release_year'] = pd.to_numeric(df['release_year'], errors = 'coerce')
df = df.dropna(subset=['release_year'])
df['release_year'] = df['release_year'].astype(int)

year_type = df.groupby(['release_year', 'type']).size().reset_index(name='count')

pivot_df = year_type.pivot(index='release_year', columns = 'type', values='count')

plt.figure(figsize=(12,6))
plt.plot(pivot_df.index, pivot_df['MOVIE'], marker='o', label='Movies')
plt.plot(pivot_df.index, pivot_df['SHOW'], marker='o', label='TV,Shows')

plt.title("chart - 4: Release Trend of Movies vs TV Shows", fontsize=14)
plt.xlabel("Release Year", fontsize=12)
plt.ylabel("Number of Titles Released", fontsize=12)
plt.xticks(rotation=60)
plt.grid(alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
````
